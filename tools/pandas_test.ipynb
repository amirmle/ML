### ORDERING AND GROUPING

```python
df = copy.copy(main)
df.loc[0, 'city'] = 'Tehran'
display(df.sort_values(by='age', ascending=False))

display(df.groupby('city')['age'].mean())


display(df.groupby('city').agg({'age':['mean','median','max','min','count']}))
display(df.agg({'age':['mean','median','max','min','count']}))
```
